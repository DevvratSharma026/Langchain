### Messages

Messages are the fundamental unit of contxt for models in Langchain. They represent the input and output of models, carrying both the content and metadata needed to represent the stat of a conversation when interacting with an LLM. Messages are objects that contain:

    1. Role - Identifies the messages type (e.g. system, user)
    2. Content - Represent the actual content of the message (like text, images, audio, docs, etc)
    3. Meta data - Optional fields such as reponse information, message IDs and token usage.

Langchain provides a standard message type that works across all model providers, ensuring consistent behaviour regardless of the model being called.

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:openai/gpt-oss-20b")
```

```python
# Human message
model.invoke("how do humans talk?")
```

### Text Prompt
Text prompt are strings - ideal for straightforward generation tasks where you do not need to retain conversation history.

```python
# we do not provide any thing specifically
model.invoke("what is langchain")
```

### Message prompt
Alternatively, you can pass in a list of messages to the model by providing a list of message objects.
Refer to NOTION NOTEBOOK

```python
# import all the message types
from langchain.messages import SystemMessage, HumanMessage, AIMessage
messages = [
    SystemMessage("You are a expert standup comedian"),
    HumanMessage("Write a script on AI")
]

response=model.invoke(messages)
response.content
```

```python
system_mesg = SystemMessage("You are a helpful coding assistant.")

messages = [
    system_mesg,
    HumanMessage("How do i create a REST API")
]

response = model.invoke(messages)
response.content
```

```python
## Detailed info to the LLM through message it will be more detailed and practical
system_mesg = SystemMessage("You are a Senior python developer with 10+ years of experienc in building backend technologies.")

messages = [
    system_mesg,
    HumanMessage("How do i create a REST API")
]

response = model.invoke(messages)
response.content
```

```python
#Maually adding AI message

#create an AI message
ai_msg = AIMessage("I'd be happy to help you with that quetion!")

#Add conversation
messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    ai_msg,
    HumanMessage("Great! what's 2+2")
]

response = model.invoke(messages)
print(response.content)
```

```python
from langchain_core.messages import AIMessage, HumanMessage, ToolMessage

# 1. Corrected AIMessage with 'tool_calls' (plural)
ai_msg = AIMessage(
    content="",
    tool_calls=[
        {"name": "get_weather", "args": {"location": "New Delhi"}, "id": "call_123"}
    ],
)

# 2. Execute tool and create result message
weather_result = "sunny 34 degree"
tool_mesg = ToolMessage(
    content=weather_result,
    tool_call_id="call_123",  # must match the call ID exactly
)

# 3. Continue conversation
messages = [
    HumanMessage("What's the weather in New Delhi"),
    ai_msg,
    tool_mesg,
]

# 4. Invoke model
response = model.invoke(messages)  # or model.invoke(messages) depending on your variable name
print(response)
```
