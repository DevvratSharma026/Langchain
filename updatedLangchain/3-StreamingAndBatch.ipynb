### Streaming and Batch

### Streaming

Most models can stream their output content while it is beign generated. By dispalying output progressively, streaming significantly impoves user experence, paritcularly for longer responses. Calling stream() returns an iteratior that yields output checks as they are produced. YOu can use a loop to process each chink in real-time

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
model.stream("write a poem about the beauty of nature in 100 words")
```

```python
for chunk in model.stream("write a poem about the beauty of nature in 100 words"):
    print(chunk.text, end="", flush=True)
```

### Batch

Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the porcessing can be done in parallel.

```python
responses = model.batch_as_completed([
    "Why is the sky blue?",
    "how the airplane flies?",
    "What is the meaning of life?"
],
    config={
        'max_concurrency':5 # This is the maximum number of concurrent requests to the model. You can adjust this value based on your needs and the capabilities of the model.
    }                        
)

for response in responses:
    print(response)
```
